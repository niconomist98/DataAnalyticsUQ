```python
import pandas as  pd
```

```python
#Leyendo la tabla de caracteristicas generales
ft = spark.read.format("delta").table("workspace.electiva.ft")
desempleados = spark.read.format("delta").table("workspace.electiva.desempleados")
cgse = spark.read.format("delta").table("workspace.electiva.cgse")

# Convertir el DataFrame de Spark a Pandas
ft = ft.toPandas()
desempleados = desempleados.toPandas()
cgse = cgse.toPandas()

```

```python
#Primero hay que anexar la p3271 a la tabla de desempleados
desempleados = pd.merge(desempleados, cgse[['DIRECTORIO', 'SECUENCIA_P', 'ORDEN', 'P3271','P6040']], 
                        on=['DIRECTORIO', 'SECUENCIA_P', 'ORDEN'], how='left')
```

```python
## Calcular cantidad de personas desempleados en el quindio en estos segmentos de edad 
#  -18 a 24
# - 24 a 30
#- 30 a 40
#- 40 a 50
#- 60 en adelante


# 1.Cantidad de personas  desempleados en el quindio entre los 18 y los 24 años 
D_18_24 = desempleados[(desempleados['P6040']>=18) & (desempleados['P6040']<24) & (desempleados['DSI']==1) & (desempleados['DPTO']==63)]['FEX_C18'].sum()
print(f"en el quindio hay {D_18_24} desempleados en el quindio entre los 18 y los 24 años ")

# 2.Cantidad de personas  desempleados en el quindio entre los 24 a 30
D_24_30 = desempleados[(desempleados['P6040']>=24) & (desempleados['P6040']<30) & (desempleados['DSI']==1) & (desempleados['DPTO']==63)]['FEX_C18'].sum()
print(f"en el quindio hay {D_24_30} desempleados en el quindio entre los 24 a 30 años ")

# 3.Cantidad de personas  desempleados en el quindio entre los 30 a 40
D_30_40 = desempleados[(desempleados['P6040']>=30) & (desempleados['P6040']<40) & (desempleados['DSI']==1) & (desempleados['DPTO']==63)]['FEX_C18'].sum()
print(f"en el quindio hay {D_30_40} desempleados en el quindio entre los 30 a 40 ")

# 4.Cantidad de personas  desempleados en el quindio entre los 40 a 50
D_40_50 = desempleados[(desempleados['P6040']>40) & (desempleados['P6040']<=50) & (desempleados['DSI']==1) & (desempleados['DPTO']==63)]['FEX_C18'].sum()
print(f"en el quindio hay {D_40_50} desempleados en el quindio entre los 40 a 50")

# 5.Cantidad de personas  desempleados en el quindio de 60 en adelante
D_60_ad = desempleados[(desempleados['P6040']>60) & (desempleados['DSI']==1) & (desempleados['DPTO']==63)]['FEX_C18'].sum()
print(f"en el quindio hay {D_60_ad} desempleados en el quindio de 60 en adelante")
```

```python
## Primero hay que añadir la P3271 al modulo de fuerza de trabajo 
ft = pd.merge(ft, cgse[['DIRECTORIO', 'SECUENCIA_P', 'ORDEN', 'P3271', 'P6040']],
                       on=['DIRECTORIO', 'SECUENCIA_P', 'ORDEN'], how='left')
```

```python
## Calcular PEA en el quindio en estos segmentos de edad 
#  -18 a 24
# - 24 a 30
#- 30 a 40
#- 40 a 50
#- 60 en adelante

# 1.Cantidad de personas  ft en el quindio entre los 18 y los 24 años 
PEA_18_24 = ft[(ft['P6040']>=18) & (ft['P6040']<24) & (ft['FT']==1) & (ft['DPTO']==63)]['FEX_C18'].sum()
print(f"en el quindio hay {PEA_18_24} pea en el quindio entre los 18 y los 24 años ")

# 2.Cantidad de personas  ft en el quindio entre los 24 a 30
PEA_24_30 = ft[(ft['P6040']>=24) & (ft['P6040']<30) & (ft['FT']==1) & (ft['DPTO']==63)]['FEX_C18'].sum()
print(f"en el quindio hay {PEA_24_30} pea en el quindio entre los 24 a 30 años ")

# 3.Cantidad de personas  ft en el quindio entre los 30 a 40
PEA_30_40 = ft[(ft['P6040']>=30) & (ft['P6040']<40) & (ft['FT']==1) & (ft['DPTO']==63)]['FEX_C18'].sum()
print(f"en el quindio hay {PEA_30_40} pea en el quindio entre los 30 a 40 ")

# 4.Cantidad de personas  ft en el quindio entre los 40 a 50
PEA_40_50 = ft[(ft['P6040']>40) & (ft['P6040']<=50) & (ft['FT']==1) & (ft['DPTO']==63)]['FEX_C18'].sum()
print(f"en el quindio hay {PEA_40_50} pea en el quindio entre los 40 a 50")

# 5.Cantidad de personas  ft en el quindio de 60 en adelante
PEA_60_ad = ft[(ft['P6040']>60) & (ft['FT']==1) & (ft['DPTO']==63)]['FEX_C18'].sum()
print(f"en el quindio hay {PEA_60_ad} pea en el quindio de 60 en adelante")
```

```python
# Tasa de desempleo por grupos de edad
TDQ_18_24= D_18_24/PEA_18_24*100
print(f"la tasa de desempleo en el quindio entre los 18 y los 24 años es {TDQ_18_24}%")

TDQ_24_30=D_24_30/PEA_24_30*100
print(f"la tasa de desempleo en el quindio entre los 24 a 30 años es {TDQ_24_30}%")     

TDQ_30_40=D_30_40/PEA_30_40*100
print(f"la tasa de desempleo en     el quindio entre los 30 a 40 años es {TDQ_30_40}%")

TDQ_40_50=D_40_50/PEA_40_50*100 
print(f"la tasa de desempleo en el quindio entre los 40 a 50 años es {TDQ_40_50}%") 

TDQ_60_ad=D_60_ad/PEA_60_ad*100 
print(f"la tasa de desempleo en el quindio de 60 en adelante es {TDQ_60_ad}%")
```
